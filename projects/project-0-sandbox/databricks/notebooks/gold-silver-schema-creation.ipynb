```python
%sql
CREATE TABLE IF NOT EXISTS silver.fact_registrations
USING DELTA
LOCATION 'abfss://processed@stregistrationsde001.dfs.core.windows.net/reports/fact_registrations';
```

```python
%sql
SELECT COUNT(*) FROM silver.fact_registrations;
```

```python
%sql
CREATE OR REPLACE VIEW gold.v_daily_registrations AS
SELECT
  load_type,
  unit_name,
  COUNT(*) AS registrations
FROM silver.fact_registrations
GROUP BY load_type, unit_name;
```

```python
%sql
SELECT *
FROM gold.v_daily_registrations
ORDER BY load_type, unit_name;
```
